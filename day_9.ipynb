5. Write a function called check-season, it takes a month parameter and returns the season: Autumn, Winter, Spring or Summer.


```python
# Check the season
def check_season(month):
    month = month.strip().lower()

    seasons = {
        "autumn": ["september", "october", "november"],
        "winter": ["december", "january", "february"],
        "spring": ["march", "april", "may"],
        "summer": ["june", "july", "august"]
    }

    for season, months in seasons.items():
        if month in months:
            return season.capitalize()

    return "Invalid month"

print(check_season("March"))  
```

6. Write a function called calculate_slope which return the slope of a linear equation


```python
#Calculate the slope of a linear equation
def calculate_slope(x1, y1, x2, y2):
    if x2 == x1:
        raise ValueError("The slope is undefined for a vertical line")

    return (y2 - y1) / (x2 - x1)

print(calculate_slope(2, 3, 4, 7))  
```

7. Quadratic equation is calculated as follows: ax² + bx + c = 0. Write a function which calculates solution set of a quadratic equation, solve_quadratic_eqn.


```python
# Solve a quadratic equation: ax² + bx + c = 0
import cmath
def solve_quadratic_eqn(a, b, c):
    if a == 0:
        if b == 0:
            return [] if c != 0 else "Infinite solutions"
        return [-c / b]

    discriminant = b**2 - 4 * a * c
    root1 = (-b + cmath.sqrt(discriminant)) / (2 * a)
    root2 = (-b - cmath.sqrt(discriminant)) / (2 * a)

    if root1 == root2:
        return [root1]

    return [root1, root2]

print(solve_quadratic_eqn(1, -3, 2))
```

9. Declare a function named reverse_list. It takes an array as a parameter and it returns the reverse of the array (use loops).


```python
# Reverse a list using loops
def reverse_list(items):
    reversed_items = []

    for item in items:
        reversed_items.insert(0, item)

    return reversed_items

print(reverse_list([1, 2, 3, 4, 5]))

print(reverse_list(["A", "B", "C"]))
```

```python
#Remove an item from a list
def remove_item(items, item):
    new_list = items.copy()

    if item in new_list:
        new_list.remove(item)

    return new_list

food_staff = ["Potato", "Tomato", "Mango", "Milk"]

print(remove_item(food_staff, "Mango"))

numbers = [2, 3, 7, 9]

print(remove_item(numbers, 3))
```

14. Declare a function named sum_of_odds. It takes a number parameter and it adds all the odd numbers in that range.


```python
#Sum odd numbers within a range
def sum_of_odds(n):
    total = 0

    for number in range(1, n + 1):
        if number % 2 != 0:
            total += number

    return total

print(sum_of_odds(10)) 
```

1. Declare a function named evens_and_odds . It takes a positive integer as parameter and it counts number of evens and odds in the number.


```python
#Count even and odd numbers
def evens_and_odds(number):
    if not isinstance(number, int) or isinstance(number, bool) or number < 0:
        raise ValueError("Please provide a non-negative integer")

    evens = 0
    odds = 0

    for num in range(number + 1):
        if num % 2 == 0:
            evens += 1
        else:
            odds += 1

    print(f"The number of odds are {odds}.")
    print(f"The number of evens are {evens}.")

evens_and_odds(100)
```

2. Call your function factorial, it takes a whole number as a parameter and it return a factorial of the number


```python
#Calculate factorial
def factorial(number):
    if not isinstance(number, int) or isinstance(number, bool) or number < 0:
        raise ValueError("Please provide a non-negative whole number")

    result = 1

    for num in range(1, number + 1):
        result *= num

    return result

print(factorial(5))  
print(factorial(0))  
```

1. Write a function called is_prime, which checks if a number is prime.


```python
#Check whether a number is prime
def is_prime(number):
    if not isinstance(number, int) or isinstance(number, bool):
        return False

    if number < 2:
        return False

    for num in range(2, int(number ** 0.5) + 1):
        if number % num == 0:
            return False

    return True

print(is_prime(7))   
print(is_prime(10))  
print(is_prime(1))  
print(is_prime(13))  
```

2. Write a functions which checks if all items are unique in the list.


```python
#Check whether all items in a list are unique
def all_unique(items):
    for i in range(len(items)):
        for j in range(i + 1, len(items)):
            if items[i] == items[j]:
                return False

    return True

print(all_unique([1, 2, 3, 4, 5]))  
print(all_unique([1, 2, 2, 4]))    
print(all_unique(["A", "B", "C"])) 
```


3. Write a function which checks if all the items of the list are of the same data type.


```python
#Check whether all items have the same data type
def same_data_type(items):
    if not items:
        return True

    first_type = type(items[0])

    for item in items:
        if type(item) != first_type:
            return False

    return True

print(same_data_type([1, 2, 3, 4]))          
print(same_data_type(["A", "B", "C"]))      
print(same_data_type([1, "Python", 3]))      
print(same_data_type([True, 1]))            
```

4. Write a function which check if provided variable is a valid python variable


```python
#Check whether a variable name is valid in Python
import keyword

def is_valid_variable(variable):
    if not isinstance(variable, str):
        return False

    return variable.isidentifier() and not keyword.iskeyword(variable)

print(is_valid_variable("name"))      
print(is_valid_variable("first_name")) 
print(is_valid_variable("first-name")) 
print(is_valid_variable("123name"))    
print(is_valid_variable("for"))        
```

2. Flatten the following list of lists of lists to a one dimensional list :


```python
#Flatten a list of lists of lists
list_of_lists = [
    [[1, 2, 3]],
    [[4, 5, 6]],
    [[7, 8, 9]]
]
flattened_list = [
    number
    for outer_list in list_of_lists
    for inner_list in outer_list
    for number in inner_list
]

print(flattened_list)
```
