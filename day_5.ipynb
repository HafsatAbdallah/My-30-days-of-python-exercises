### 1. The following is a list of 10 students ages:
ages = [19, 22, 19, 24, 20, 25, 26, 24, 25, 24]
* Sort the list and find the min and max age 

* Find the median age (one middle item or two middle items divided by two) 
* Find the average age (sum of all items divided by their number ) 
* Find the range of the ages (max minus min) 
* Compare the value of (min - average) and (max - average), use abs() method 
* Add the min age and the max age again to the list 
1. Find the middle country(ies) in the countries list
2. Divide the countries list into two equal lists if it is even if not one more country for the first half.
3. ['China', 'Russia', 'USA', 'Finland', 'Sweden', 'Norway', 'Denmark']. Unpack the first three countries and the rest as scandic countries.


```python
# Sort the list and find the min and max age 
ages = [19, 22, 19, 24, 20, 25, 26, 24, 25, 24]
ages.sort()

min_age = min(ages)
max_age = max(ages)

print("Sorted ages:", ages)
print("Minimum age:", min_age)
print("Maximum age:", max_age)
```

```python
# Add the min age and the max age again to the list 
ages.append(min_age)
ages.append(max_age)

print(ages)
```

```python
# Find the median age (one middle item or two middle items divided by two) 
ages.sort()

n = len(ages)

if n % 2 == 0:
    median_age = (ages[n // 2 - 1] + ages[n // 2]) / 2
else:
    median_age = ages[n // 2]

print("Median age:", median_age)
```

```python
# Find the average age (sum of all items divided by their number ) 
average_age = sum(ages) / len(ages)

print("Average age:", average_age)
```

```python
# Find the range of the ages (max minus min) 
age_range = max(ages) - min(ages)

print("Age range:", age_range)
```

```python
# Compare the value of (min - average) and (max - average), use abs() method 
min_average_difference = abs(min_age - average_age)
max_average_difference = abs(max_age - average_age)

print("Difference between minimum age and average:", min_average_difference)
print("Difference between maximum age and average:", max_average_difference)

print(min_average_difference == max_average_difference)
```

```python
# 1. Find the middle country(ies) in the countries list from countries import countries
countries = ['China', 'Russia', 'USA', 'Finland', 'Sweden', 'Norway', 'Denmark']
middle = len(countries) // 2

if len(countries) % 2 == 0:
    print("Middle countries:", countries[middle - 1:middle + 1])
else:
    print("Middle country:", countries[middle])
```

```python
# 2. Divide the countries list into two equal lists if it is even if not one more country for the first half.
Countries = ['China', 'Russia', 'USA', 'Finland', 'Sweden', 'Norway', 'Denmark']
middle = (len(countries) + 1) // 2

first_half = countries[:middle]
second_half = countries[middle:]

print("First half:", first_half)
print("Second half:", second_half)

print("Number in first half:", len(first_half))
print("Number in second half:", len(second_half))
```

```python
# 3. ['China', 'Russia', 'USA', 'Finland', 'Sweden', 'Norway', 'Denmark']. Unpack the first three countries and the rest as scandic countries.
first, second, third, *scandic_countries = countries

print(first)
print(second)
print(third)
print(scandic_countries)
```

```python
A = {19, 22, 24, 20, 25, 26}
B = {19, 22, 20, 25, 26, 24, 28, 27}
```

```python
# Join A and B
A_union_B = A.union(B)

print(A_union_B)
```

```python
# Find A intersection B
A_intersection_B = A.intersection(B)

print(A_intersection_B)
```

```python
# Is A subset of B
print(A.issubset(B))
```

```python
# Are A and B disjoint sets
print(A.isdisjoint(B))
```

```python
# Join A with B and B with A
A.update(B)
print("A after joining B:", A)

B.update(A)
print("B after joining A:", B)
```

```python
# What is the symmetric difference between A and B
A = {19, 22, 24, 20, 25, 26}
B = {19, 22, 20, 25, 26, 24, 28, 27}

print(A.symmetric_difference(B))
```
