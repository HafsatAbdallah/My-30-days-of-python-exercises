1. Write a function which count number of lines and number of words in a text. All the files are in the data the folder: a) Read obama_speech.txt file and count number of lines and words b) Read michelle_obama_speech.txt file and count number of lines and words c) Read donald_speech.txt file and count number of lines and words d) Read melina_trump_speech.txt file and count number of lines and words


```python

def count_lines_and_words(filename):
    with open(filename, "r", encoding="utf-8") as file:
        lines = file.readlines()

    number_of_lines = len(lines)
    number_of_words = sum(len(line.split()) for line in lines)

    return number_of_lines, number_of_words


# a) Obama's speech
print("Obama:", count_lines_and_words(
    "./data/obama_speech.txt"
))

# b) Michelle Obama's speech
print("Michelle:", count_lines_and_words(
    "./data/michelle_obama_speech.txt"
))

# c) Donald Trump's speech
print("Donald:", count_lines_and_words(
    "./data/donald_speech.txt"
))

# d) Melania Trump's speech
print("Melania:", count_lines_and_words(
    "./data/melina_trump_speech.txt"
))
```


2. Read the countries_data.json data file in data directory, create a function that finds the ten most spoken languages

```python

import json
from collections import Counter


def most_spoken_languages(filename, number):
    with open(filename, "r", encoding="utf-8") as file:
        countries = json.load(file)

    language_counter = Counter()

    for country in countries:
        language_counter.update(country["languages"])

    return [
        (count, language)
        for language, count in language_counter.most_common(number)
    ]


print(most_spoken_languages(
    "./data/countries_data.json", 10
))

print(most_spoken_languages(
    "./data/countries_data.json", 3
))
```

3. Read the countries_data.json data file in data directory, create a function that creates a list of the ten most populated countries

```python

import json


def most_populated_countries(filename, number):
    with open(filename, "r", encoding="utf-8") as file:
        countries = json.load(file)

    populated_countries = []

    for country in countries:
        populated_countries.append({
            "country": country["name"],
            "population": country["population"]
        })

    populated_countries.sort(
        key=lambda country: country["population"],
        reverse=True
    )

    return populated_countries[:number]


print(most_populated_countries(
    "./data/countries_data.json", 10
))

print(most_populated_countries(
    "./data/countries_data.json", 3
))
```

4. Extract all incoming email addresses as a list from the email_exchange_big.txt file. 

```python

import re


def extract_email_addresses(filename):
    with open(filename, "r", encoding="utf-8") as file:
        text = file.read()

    email_pattern = r'[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}'

    email_addresses = re.findall(email_pattern, text)

    return email_addresses


emails = extract_email_addresses(
    "./data/email_exchange_big.txt"
)

print(emails)
print("Total email addresses:", len(emails))
```

5. Find the most common words in the English language. Call the name of your function find_most_common_words, it will take two parameters - a string or a file and a positive integer, indicating the number of words. Your function will return an array of tuples in descending order. Check the output


```python

import re
from collections import Counter
from pathlib import Path


def read_text_or_file(source):
    path = Path(source)

    try:
        if path.is_file():
            return path.read_text(encoding="utf-8")
    except (OSError, ValueError):
        pass

    return str(source)


def find_most_common_words(source, number):
    if not isinstance(number, int) or isinstance(number, bool) or number <= 0:
        raise ValueError("number must be a positive integer")

    text = read_text_or_file(source)

    words = re.findall(r"\b[\w']+\b", text)

    word_count = Counter(words)

    return word_count.most_common(number)


print(find_most_common_words("./data/sample.txt", 10))

print(find_most_common_words("./data/sample.txt", 5))
```

7. Write a python application that checks similarity between two texts. It takes a file or a string as a parameter and it will evaluate the similarity of the two texts. For instance check the similarity between the transcripts of Michelle's and Melina's speech. You may need a couple of functions, function to clean the text(clean_text), function to remove support words(remove_support_words) and finally to check the similarity(check_text_similarity). List of stop words are in the data directory


```python

import re
import math
from collections import Counter
from pathlib import Path


# Read either a file or a text string
def read_text(source):
    path = Path(source)

    try:
        if path.is_file():
            return path.read_text(encoding="utf-8")
    except (OSError, ValueError):
        pass

    return str(source)


# 1. Clean the text
def clean_text(text):
    text = text.lower()

    # Remove punctuation and retain words
    text = re.sub(r"[^a-z0-9\s]", " ", text)

    # Remove extra spaces
    text = re.sub(r"\s+", " ", text).strip()

    return text


# 2. Remove stop words
def remove_support_words(text):
    from data.stop_words import stop_words

    words = text.split()

    filtered_words = [
        word for word in words
        if word not in stop_words
    ]

    return filtered_words


# 3. Calculate cosine similarity
def check_text_similarity(text1, text2):
    text1 = clean_text(read_text(text1))
    text2 = clean_text(read_text(text2))

    words1 = remove_support_words(text1)
    words2 = remove_support_words(text2)

    counter1 = Counter(words1)
    counter2 = Counter(words2)

    all_words = set(counter1) | set(counter2)

    dot_product = sum(
        counter1[word] * counter2[word]
        for word in all_words
    )

    magnitude1 = math.sqrt(
        sum(value ** 2 for value in counter1.values())
    )

    magnitude2 = math.sqrt(
        sum(value ** 2 for value in counter2.values())
    )

    if magnitude1 == 0 or magnitude2 == 0:
        return 0.0

    similarity = dot_product / (magnitude1 * magnitude2)

    return round(similarity * 100, 2)


# Compare Michelle's and Melania's speeches
similarity = check_text_similarity(
    "./data/michelle_obama_speech.txt",
    "./data/melina_trump_speech.txt"
)

print("Text similarity:", similarity, "%")
```

9. Read the hacker news csv file and find out: a) Count the number of lines containing python or Python b) Count the number lines containing JavaScript, javascript or Javascript c) Count the number lines containing Java and not JavaScript

```python

import csv
import re


def analyze_hacker_news(filename):
    python_count = 0
    javascript_count = 0
    java_count = 0

    with open(filename, "r", encoding="utf-8", newline="") as file:
        reader = csv.reader(file)

        # Skip the header row
        next(reader, None)

        for row in reader:
            line = " ".join(row)

            # A. Contains Python
            if re.search(r"\bpython\b", line, re.IGNORECASE):
                python_count += 1

            # B. Contains JavaScript
            if re.search(r"\bjavascript\b", line, re.IGNORECASE):
                javascript_count += 1

            # C. Contains Java but not JavaScript
            if (
                re.search(r"\bjava\b", line, re.IGNORECASE)
                and not re.search(
                    r"\bjavascript\b", line, re.IGNORECASE
                )
            ):
                java_count += 1

    return {
        "Python": python_count,
        "JavaScript": javascript_count,
        "Java (excluding JavaScript)": java_count
    }


results = analyze_hacker_news("./data/hacker_news.csv")

print(results)
```
